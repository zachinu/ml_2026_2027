````python
# Accessing a file using file locks to get variable names
with open(osp.join('..','data','mia','miaPassengers.names'), 'r', encoding='utf-8-sig') as f:
    columns = f.read().strip().split(',')

columns
````

We created a pointer ```f``` to the file we want to access and attached a read lock. We also define the encoding, meaning the character space to use to interpret the content, utf8. The ```utf-8-stg``` variant is needed to support the specific BOM character of this file. Every file has a BOM and EOF character.

When reading the contents of the file we remove whitespaces using the ```strip()``` command, then we create a list with the contents of the file using the ```split()``` command since they are comma separated.

````python
df = pd.read_csv(osp.join('..','data','mia','miaPassengers.data'), names=columns)
df.info()
df.head()
````

## Case-Study 02: Climate of Malta Weather scraping

Scraping is an important data scientist tool, which should be used responsibly. Misuse could lead to being blacklisted, ssrvice interruption or even legal consequences.

We shall be accessing wikipedia which is a permissive license, and has a strong infrastructure.

````python
# Load libraries
import pandas as pd # Load data in a data frame

# For scraping there two components, requests and parsing
import requests # Allows us to do HTML protocol request
from bs4 import BeautifulSoup # The HTML parsing
````

````python
# Submitting an HTML request
url = "https://en.wikipedia.org/wiki/Climate_of_Malta"

response = requests.get(
    url,
headers={"User-Agent": "MCAST-ClassroomDemo/1.0 (educational example)"},
    timeout=1000
)
response.raise_for_status()
````

````python
type(response.content)
````

````python
# Now we need to parse the HTML response
soup = BeautifulSoup(response.content, "html.parser")
table = soup.select_one("table.wikitable")
````

````python
if table is None:
    raise ValueError("Expected table not found!")
````

````python
data = [
    [cell.get_text(" ", strip=True)
        for cell in row.find_all(["th", "td"])]
    for row in table.find_all("tr")
]
data
````

````python
df = pd.DataFrame(data[1:], columns=data[0])
df
````
